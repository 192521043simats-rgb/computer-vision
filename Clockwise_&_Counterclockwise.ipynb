```python
# Rotation of an Image (Clockwise and Counter Clockwise)

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get uploaded file name
image_path = next(iter(uploaded))

# Read image
image = cv2.imread(image_path)

# Convert BGR to RGB for display
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Rotate Clockwise
clockwise = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)

# Rotate Counter Clockwise
counter_clockwise = cv2.rotate(image, cv2.ROTATE_90_COUNTERCLOCKWISE)

# Display Images
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(clockwise)
plt.title("Clockwise Rotation")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(counter_clockwise)
plt.title("Counter Clockwise Rotation")
plt.axis("off")

plt.show()
```
