```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get file name
image_path = next(iter(uploaded))

# Read image
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Get original size
height, width = image.shape[:2]

# Bigger image (2x)
bigger = cv2.resize(image, (width * 2, height * 2), interpolation=cv2.INTER_LINEAR)

# Smaller image (Half size)
smaller = cv2.resize(image, (width // 2, height // 2), interpolation=cv2.INTER_AREA)

# Original Image
plt.figure(figsize=(5,5))
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()

# Bigger Image
plt.figure(figsize=(10,10))
plt.imshow(bigger)
plt.title("Bigger Image")
plt.axis("off")
plt.show()

# Smaller Image
plt.figure(figsize=(3,3))
plt.imshow(smaller)
plt.title("Smaller Image")
plt.axis("off")
plt.show()
```
