```python
# Import libraries
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get the uploaded file name
image_path = list(uploaded.keys())[0]

# Read the image
img = cv2.imread(image_path)

# Convert image to grayscale
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Display original image
print("Original Image:")
cv2_imshow(img)

# Display grayscale image
print("Grayscale Image:")
cv2_imshow(gray_img)

# Save grayscale image
cv2.imwrite("gray_image.jpg", gray_img)

print("Grayscale image saved as gray_image.jpg")
```

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Choose an image file
uploaded = files.upload()
filename = next(iter(uploaded))

# Read the image
image = cv2.imread(filename)

# Apply Gaussian Blur
blur = cv2.GaussianBlur(image, (5, 5), 10)

# Display Original Image
print("Original Image:")
cv2_imshow(image)

# Display Blurred Image
print("Blurred Image:")
cv2_imshow(blur)

# Save the blurred image
cv2.imwrite("blurred_image.jpg", blur)

print("Blurred image saved as 'blurred_image.jpg'")
```

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Choose an image file
uploaded = files.upload()
filename = next(iter(uploaded))

# Read the image
img = cv2.imread(filename)

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Canny Edge Detection
edges = cv2.Canny(gray, 100, 200)

# Display Original Image
print("Original Image:")
cv2_imshow(img)

# Display Edge Detected Image
print("Canny Edge Detection:")
cv2_imshow(edges)

# Save the output image
cv2.imwrite("Canny_Edges_3rd.jpg", edges)

print("Edge detected image saved as 'Canny_Edges_3rd.jpg'")
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Choose an image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read original image in color
original = cv2.imread(filename)

# Convert to grayscale for dilation
gray = cv2.cvtColor(original, cv2.COLOR_BGR2GRAY)

# Create kernel
kernel = np.ones((5,5), np.uint8)

# Dilate image
dilated_img = cv2.dilate(gray, kernel, iterations=1)

# Show original image
print("Original Image:")
cv2_imshow(original)

# Show dilated image
print("Dilated Image:")
cv2_imshow(dilated_img)

# Save image
cv2.imwrite("Dilated_Image.jpg", dilated_img)

print("Dilated image saved as 'Dilated_Image.jpg'")
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Choose an image file
uploaded = files.upload()
filename = next(iter(uploaded))

# Read the original image
img = cv2.imread(filename)

# Convert to grayscale
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Create a 5x5 rectangular kernel
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# Apply erosion
eroded_image = cv2.erode(gray_image, kernel, iterations=1)

# Display Original Image
print("Original Image:")
cv2_imshow(img)

# Display Eroded Image
print("Eroded Image:")
cv2_imshow(eroded_image)

# Save the eroded image
cv2.imwrite("eroded_image.jpg", eroded_image)

print("Eroded image saved as 'eroded_image.jpg'")
```
